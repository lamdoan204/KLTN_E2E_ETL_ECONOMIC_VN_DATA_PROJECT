```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
base_url= 'https://www.nso.gov.vn/bao-cao-tinh-hinh-kinh-te-xa-hoi-hang-thang/'
```

```python
page = 1
flag = True
titles = []
links = []
while flag: 
    if page == 1: 
        url = base_url
    else:
        url = base_url+ f'?paged={page}'

    print(f'Crawling Page: {page} ...........................')

    res = requests.get(url, verify=False)
    soup = BeautifulSoup(res.text, "html.parser")
    container = soup.find('div', class_= 'archive-container')
    the_a = container.find_all('a', class_= None)
    the_h3 = container.find_all('h3', class_=None)
    
    # Lấy title bài báo cáo
    this_page_titles = []
    for x in the_h3:
        t = x.get_text(strip= True)
        if '2024' in t:
            flag = False
            break
        this_page_titles.append(t)
        
    # lấy link bài báo cáo 
    this_page_links = []
    for x in the_a:
        t = x['href']
        if '2024' in t: break
        this_page_links.append(t)

    titles += this_page_titles
    links += this_page_links
    
    page += 1    
```

```python
# vào từng link tải file excel về

for i in range(len(titles)):
    # vào từng file
    title =  str.lower(titles[i])
    link = links[i]
    # loại bỏ những bài không phải báo cáo tình hình kinh tế
    print(title)
    if 'báo cáo tình hình kinh tế' not in title: 
        continue
    res = requests.get(link, verify=False)
    soup = BeautifulSoup(res.text, 'html.parser')

    # lấy lấy link file excel
    excel_url = ''
    for a in soup.find_all('a', href=True):
        href = a['href']
        if href.endswith(('.xls', '.xlsx')):
            excel_url = href
    files_path = '../Bronze_Excel_Financial_Report'

    # phân loại file
    # phân theo năm
    year = title[len(title)-4::]
    files_path += f'/{year}'
    if 'quý' in title:
        files_path += '/Is_Quy/'
    else:
        files_path += '/IS_Thang/'
    
    os.makedirs(files_path, exist_ok=True)
    
    file_name = excel_url.split('/')[-1]
    files_path += file_name

    excel_file = requests.get(excel_url, verify= False)
    with open(files_path, "wb") as f:
        f.write(excel_file.content)
```
